```python
import matplotlib.pyplot as plt
import CPG_approximation as cpg
import numpy as np
```

```python

T = cpg.Test()

cycles = 4
theta = 0
theta_d = 0.2
amp = 1.5
offset = 0

steps = (1/theta_d) * (1 /T.TimeStep) * cycles * np.pi *2
i = 0

xs = []
ys = []
while i < steps:
    p = T.get_foot_position(theta, amp, offset);
    xs.append(p[0])
    ys.append(p[1])
    theta += theta_d * T.TimeStep
    i+=1

fig = plt.figure(figsize=(8, 3.5))
fig.tight_layout()
plt.title("Example foot trajectory")
plt.xlabel("relative x-position")
plt.ylabel("relative y-position")
plt.xlim(-0.08, 0.08)
# plt.ylim(-0.195, -0.135)
plt.gca().set_aspect('equal')
plt.plot(xs, ys);
plt.hlines(-0.18, -0.075, 0.075, linestyles="dashed", label="dStep", colors="grey")
plt.vlines(0, -0.19, -0.18, label="gP", colors="red", linestyles="dashed")
plt.vlines(0, -0.18, -0.14, label="gC", colors="orange", linestyles="dashed")
plt.legend(loc='upper center', bbox_to_anchor=(1.1, 1), ncol=1)

fig.subplots_adjust(right=0.85)
plt.margins(0.1, 0.1)
# Reduce whitespace around the plotting area
fig.subplots_adjust(right=0.85)

fig.savefig("foot_trajectory.svg", format="svg")
```
